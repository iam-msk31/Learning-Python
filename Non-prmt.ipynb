### -> Accessing Elements

```python
a=[10,20,30,40]
print(a[0])#10
print(a[1])#20
print(a[2])#30
print(a[3])#40

print(a[-1])#40
print(a[-2])#30
print(a[-3])#20
print(a[-4])#10

print(a[1:4])#[20,30,40]
print(a[ :4])#[10,20,30,40]
print(a[1:])#[20, 30, 40]
```

### -> Append 

```python
b = [1,2,3,4]
print("before append ;",b)
b.append("five")#add "five in last element"
b.append(["six","seven"])
print("after append",b)
```

### -> Extend

```python
b = [1,2,3,4]
print("before extend ;",b)
b.extend(["five","six","seven"])# After extend [1, 2, 3, 4, 'five', 'six', 'seven']
print("After extend" ,b)
```

### -> Insert

```python
f = ["java","Python","Js","HTML"]
print(f)
f.insert(0 , "Machine learning")#Insert Machine learning at index = 0
print(f)
```

### Count

### Sort

```python
a = [40, 10, 30, 20]

a.sort()

print(a)
```

### Reverse of sort

```python
a = [40, 10, 30, 20]

a.sort(reverse=True)

print(a)
```

### Reverse

```python
a = [1, 2, 3, 4]

a.reverse()

print(a)
```

### Copying a list

```python
a = [1, 2, 3]

b = a.copy()

print(b)
```

### Minimum value in list

```python
a = [10, 20, 30, 40]

print(min(a))
```

### Maximum value in list

```python
a = [10, 20, 30, 40]

print(max(a))
```

### Slicing

```python
a = [10, 20, 30, 40, 50]
print(a[1:4]) # [20, 30, 40]
print(a[:3]) # [10, 20, 30]
print(a[2:]) # [30, 40, 50]
print(a[::-1]) # [50, 40, 30, 20, 10]
```

### POP

```python
a = [10, 20, 30]
x = a.pop()
print(x) # 30
print(a) # [10, 20]
```

### Clear

```python
a = [1, 2, 3]
a.clear()
print(a)
```
